```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob

# Set visualization theme
sns.set_theme(style="whitegrid")

# Sample social media dataset
data = {
    'text': [
        "I absolutely love the new features on this app! Best update ever.",
        "Terrible experience. It keeps crashing every time I try to log in.",
        "The app is okay, nothing special, but it gets the job done.",
        "Mind-blowing performance and design! Highly recommend to everyone.",
        "Worst update ever. It lost all my saved data and settings.",
        "Pretty decent quality for the price. Satisfied with my order overall.",
        "Disappointed with the customer support. No one responded to my emails.",
        "Amazing experience! Totally worth every penny."
    ]
}

df = pd.DataFrame(data)

# Define sentiment functions
def get_polarity(text):
    return TextBlob(str(text)).sentiment.polarity

def get_sentiment_category(polarity):
    if polarity > 0.1:
        return 'Positive'
    elif polarity < -0.1:
        return 'Negative'
    else:
        return 'Neutral'

# Apply sentiment analysis
df['polarity'] = df['text'].apply(get_polarity)
df['sentiment'] = df['polarity'].apply(get_sentiment_category)

# Visualize the sentiment breakdown without warnings
plt.figure(figsize=(8, 5))
ax = sns.countplot(data=df, x='sentiment', hue='sentiment', order=['Positive', 'Neutral', 'Negative'], palette='Set2', legend=False)
plt.title('Social Media Sentiment Analysis', fontsize=14, fontweight='bold')
plt.xlabel('Sentiment Category', fontsize=12)
plt.ylabel('Count', fontsize=12)

# Add data labels on bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                xytext=(0, 5), 
                textcoords='offset points')

plt.tight_layout()
plt.show()
```

### Step 4: Summary of Sentiment Analysis & Insights

After analyzing and visualizing social media text data for sentiment patterns, the following conclusions were derived:

1. **Overall Public Attitude:**
   - The sentiment distribution successfully highlights the ratio of positive vs. negative brand/app perception, showing how public opinion leans across the sampled feedback.

2. **Polarity Analysis:**
   - Polarity scores effectively quantified emotional weight, separating enthusiastic customer endorsements from critical complaints.

3. **Business Value / Actionable Insights:**
   - Tracking sentiment patterns allows companies to quickly identify consumer pain points, measure marketing campaign reception, and prioritize product improvements based on real-time social feedback.
